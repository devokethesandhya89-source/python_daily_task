```python

# create a 2x4 grid
fig,axes=plt.subplots(2,4, figsize =(20,10))

# title of figure
fig.suptitle(
    "Categorical plot All types",
             fontsize=18,
             fontweight="bold"
            )
axes=axes.flatten()    # multidiamantion data convet into singel diamention data
# strip plot 1:
sns.stripplot(data = tips,
              x = "day",
              y = "total_bill",
              hue = "sex",
              palette= "dark",
              ax=axes[0]   # make sure to specify which axis to draw on
             )
# swarm plot2:
sns.swarmplot(data =tips,
              x ="day",
              y ="total_bill",
              hue ="sex",
              palette ="dark",
              ax=axes[1]
             )
# box plot 3:
sns.boxplot(data = tips,
              x ="day",
              hue ="sex",
              palette="dark",
              ax=axes[2]
             )
#  violin plot 4:
sns.violinplot(data = tips,
              x ="day",
              y ="total_bill",
              hue ="sex",
              palette="dark",
              ax=axes[3]
             )
# point plot 5:
sns.pointplot(data = tips,
              x ="day",
              y ="total_bill",
              hue ="sex",
              palette="dark",
              ax=axes[4]
             )
# bar plot 6:
sns.barplot(data = tips,
              x ="day",
              y ="total_bill",
              hue ="sex",
              palette="dark",
              ax=axes[5]
             )
# count plot 7:
sns.countplot(data = tips,
              x ="day",
              hue ="sex",
              palette="dark",
              ax=axes[6]
             )
# boxen plot 8:
sns.boxenplot(data = tips,
              x ="day",
              y ="total_bill",
              hue ="sex",
              palette="dark",
              ax=axes[7]
             )
plt.tight_layout()
plt.show()
```

# relplot()

```python
# what is relplot()function
# relplot is figure-level-function for visualization statistical relationship-creates scatter or line plots with facesing!
```

```python
# relplot with scatter
sns.relplot(data =tips,
            x='total_bill',
            y='tip',
            hue='sex',           #color by gender
            size='size',          #dot size by party size
            style='smoker',       # shape by smoker
            palette=['blue','red'],
            sizes=(20,200),      #size range
            alpha=0.8,
            height=6,
            aspect=1.3          #incresing value stresh decrease the value shrink the data
           )
```

# regplot()

```python
# it is very imp plot
# it draw a stright line between the scatter plot 
# regplot()regression line draw the bestfit line mx+c m:slop of line c:intersecting point on y
sns.regplot(data =tips,
            x='total_bill',
            y='tip',
            color ='purple',
            scatter_kws={'alpha':0.9,
                         'color':'purple'},
            line_kws={'color':'red',
                      'linewidth':5,
                      'alpha':0.9},
           )

```
