```python
import seaborn as sns
import matplotlib .pyplot as plt
import pandas as pd
import numpy as np
```

```python
# 1 import taitanic
titanic = sns.load_dataset("titanic")
titanic
```

```python
sns.catplot(data = titanic,
            x ="sex",
            hue="survived",
            kind ="count"
           )
plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("number of pasengers")
plt.show()
```

```python
sns.catplot(data = titanic,
            x = "class",
            kind = "count"
           )
plt.title("Number of passengers in class")
plt.xlabel=("passenger class")
plt.ylabel=("number of passengers")
```

```python
sns.catplot(data = titanic,
            x="class",
            hue="survived",
            kind="count"
           )
plt.title("Survival count of passenger class")
plt.xlabel("passenger class")
plt.ylabel("number of passenger")
plt.show()
```

```python

# Calculate survival percentage by gender
survival_rate = titanic.groupby("sex")["survived"].mean() * 100

# Convert to DataFrame
survival_rate = survival_rate.reset_index()

# Create categorical plot
sns.catplot(
    data=survival_rate,
    x="sex",
    y="survived",
    kind="bar"
)

plt.title("Survival Percentage by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Percentage (%)")

plt.show()
```

```python
sns.catplot(data = titanic,
            x ="sex",
            y="age",
            kind="box"
           )
plt.title("Age Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.show()
```

```python
sns.catplot(data = titanic,
            x ="class",
            y ="age",
            kind ="box"
           )
plt.title("Age Distribution by Passenger Class")
plt.xlabel(" passenger Class")
plt.ylabel("Age")
plt.show()
```

```python
sns.catplot(data = titanic,
            x ="class",
            y ="fare",
            kind="box",
            color ="skyblue"
           )
plt.title("Fare by Passenger Class")
plt.xlabel("passenger class")
plt.ylabel("fare")
plt.show()
```

```python
import numpy as np

# Create age groups
conditions = [
    titanic["age"] < 18,
    titanic["age"].between(18, 59),
    titanic["age"] >= 60
]

choices = ["Child", "Adult", "Senior"]

titanic["age_group"] = np.select(
    conditions,
    choices,
    default="unknown"
)

# Survival by age group
sns.catplot(
    data=titanic,
    x="age_group",
    hue="survived",
    kind="count"
)

plt.title("Survival by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
# Load Tips dataset
tips = sns.load_dataset("tips")

# Create categorical visualization
g = sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    col="time",
    kind="box",
    palette="Set2",
    height=5,
    aspect=1
)

# Title
g.fig.suptitle(
    "Total Bill by Day, Gender and Time",
    fontsize=16,
    fontweight="bold"
)

# Axis labels
g.set_axis_labels("Day", "Total Bill ($)")

# Legend
g.set_titles("{col_name}")
g._legend.set_title("Gender")

# Adjust title position
g.fig.subplots_adjust(top=0.82)

plt.show()
```
