```python
#3
df = df[df.isnull().sum(axis=1) <= 3]
print("Rows remaining:",len(df))
```

```python
# duplicates
duplicates = df.duplicated().sum()
duplicates
```

```python
# occurence of duplicate id
duplicates =df['Customer_ID'].duplicated().sum()
df = df.drop_duplicates(subset="Customer_ID	",keep="first")
duplicates
```

```python
#4
df['Age'] = pd.to_numeric(df['Age'],errors="coerce")
failed =df['Age'].isna().sum()
failed
```

```python
#7
Q1 =df['Age'].quantile(0.25)
Q3 =df['Age'].quantile(0.75)
IQR = Q3 - Q1
lower =Q1 -1.5 * IQR
upper =Q3 +1.5 * IQR
outlier = df[(df['Age'] < lower) | (df['Age'] > upper)]
print("Outlier :",len(outlier))

df =df[(df['Age'] <= lower) | (df['Age'] >= upper)]
```

```python
#8
median_Age =df['Age'].median()
df['Age'] =df['Age'].fillna(median_Age)
print("missing Age:",df['Age'].isna().sum())
```

```python
# clean the salary col
df['Salary'] =(df['Salary'].astype(str).str.replace(r"[$]","",regex =True))
df['Salary'] =pd.to_numeric(df['Salary'],errors ="coerce")
non_numeric =df['Salary'].isna().sum()
print("replaced value:",non_numeric)
```
