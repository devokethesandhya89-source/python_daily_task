```python
import re

def clean_age(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    match = re.search(r'\d+(\.\d+)?', value)

    if match:
        return float(match.group())

    return np.nan


# Clean age
df["age"] = df["age"].apply(clean_age)

# Convert to numeric
df["age"] = pd.to_numeric(
    df["age"],
    errors="coerce"
)

# Display result
print(df["age"].head(10))

# Check datatype
print("Data type:", df["age"].dtype)
```

```python

def clean_price(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    # Handle missing values
    if value in ["n/a", "na", "none", "null", ""]:
        return np.nan

    # Remove currency symbols and INR
    value = value.replace("₹", "")
    value = value.replace("$", "")
    value = value.replace("inr", "")
    value = value.replace(",", "")
    value = value.strip()

    # Handle values like 10k, 15K
    if value.endswith("k"):
        number = value[:-1].strip()

        try:
            return float(number) * 1000
        except ValueError:
            return np.nan

    # Extract numeric value
    match = re.search(r"\d+(\.\d+)?", value)

    if match:
        return float(match.group())

    return np.nan


# Apply cleaning
df["unit_price"] = df["unit_price"].apply(clean_price)

# Convert to numeric datatype
df["unit_price"] = pd.to_numeric(
    df["unit_price"],
    errors="coerce"
)

# Display results
print(df["unit_price"].head(10))

# Check datatype
print("\nData type:")
print(df["unit_price"].dtype)
```

```python

# 1. Calculate Q1, Q3, and IQR
Q1 = df['unit_price'].quantile(0.25)
Q3 = df['unit_price'].quantile(0.75)
IQR = Q3 - Q1

# 2. Define upper and lower bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# 3. Detect outliers before treatment
outliers_before = df[(df['unit_price'] < lower_bound) | (df['unit_price'] > upper_bound)]
print(f"Outliers Before Treatment: {len(outliers_before)}")

# 4. Apply Capping (Winsorization)
df_capped = df.copy()
df_capped['unit_price'] = np.clip(df_capped['unit_price'], lower_bound, upper_bound)

# 5. Detect outliers after capping
outliers_after = df_capped[(df_capped['unit_price'] < lower_bound) | (df_capped['unit_price'] > upper_bound)]
print(f"Outliers After Capping: {len(outliers_after)}")
```

```python
DataFrame = df
import re

# Dictionary for common number words
word_to_num = {
    'zero': 0, 'one': 1, 'two': 2, 'three': 3, 'four': 4,
    'five': 5, 'six': 6, 'seven': 7, 'eight': 8, 'nine': 9,
    'ten': 10, 'eleven': 11, 'twelve': 12, 'thirteen': 13,
    'fourteen': 14, 'fifteen': 15, 'sixteen': 16, 'seventeen': 17,
    'eighteen': 18, 'nineteen': 19, 'twenty': 20, 'thirty': 30,
    'forty': 40, 'fifty': 50, 'sixty': 60, 'seventy': 70,
    'eighty': 80, 'ninety': 90, 'hundred': 100
}

def clean_units(val):
    if pd.isna(val):
        return np.nan
    
    val_str = str(val).strip().lower()
    
    # 1. Handle explicit N/A or empty values
    if val_str in ['n/a', 'na', 'none', 'null', '', 'nan']:
        return np.nan
    
    # 2. Convert text words to numbers
    if val_str in word_to_num:
        return word_to_num[val_str]
    
    # 3. Extract numeric digits using regular expressions
    match = re.search(r'\d+(\.\d+)?', val_str)
    if match:
        return float(match.group())
    
    return np.nan

# Apply cleaning logic
df['units_sold_cleaned'] = df['units_sold'].apply(clean_units)

# Convert column to Pandas nullable integer type
df['units_sold_cleaned'] = df['units_sold_cleaned'].astype('Int64')
```

```python
def clean_discount(val):
    if pd.isna(val):
        return np.nan
    
    val_str = str(val).strip().lower()
    
    # 1. Handle no-discount and missing text indicators
    if val_str in ['none', 'no discount', '0', '0%']:
        return 0.0
    if val_str in ['n/a', 'na', 'null', '', 'nan']:
        return 0.0  # Set to np.nan if missing values should be distinguished from 0%
    
    # 2. Extract numeric digits (integers or decimals)
    match = re.search(r'\d+(\.\d+)?', val_str)
    if match:
        num = float(match.group())
        # Convert decimal fractions (e.g., 0.20 -> 20.0)
        if 0 < num < 1.0:
            num = num * 100.0
        return num
    
    return np.nan

# Apply function to create clean numeric column (0 to 100 scale)
df['discount_%'] = df['discount_amount'].apply(clean_discount)
```
