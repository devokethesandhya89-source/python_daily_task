```python
df["revenue"]=df["revenue"].fillna(df["revenue"].mean())  #fill numerical missing values with mean
```

```python
df["cost"]=df["cost"].fillna(df["cost"].median()) #fill numerical missing values with mwdian
```

```python
df["city"] = df["city"].fillna(df["city"].mode()[0])# fill with mode categorical missing value
```

```python
df = df.dropna(subset=["customer_id","order_date"]) #remove rows with missing imp values
```
