```python
# task
#class 1:circle-->method:darw , area , [attributr :(radius)]
#class 2:rectangle
#attribute :width ,hight
#method : draw,area

#class 3:triangle
#attribute:base, hight
#method:draw,area



import math

# Class 1: Circle
class Circle:

    def __init__(self, radius):
        self.radius = radius

    def draw(self):
        print("Drawing a Circle")

    def area(self):
        return math.pi * (self.radius ** 2)


# Class 2: Rectangle
class Rectangle:

    def __init__(self, width, height):
        self.width = width
        self.height = height

    def draw(self):
        print("Drawing a Rectangle")

    def area(self):
        return self.width * self.height


# Class 3: Triangle
class Triangle:

    def __init__(self, base, height):
        self.base = base
        self.height = height

    def draw(self):
        print("Drawing a Triangle")

    def area(self):
        return 0.5 * self.base * self.height


# Create objects
circle = Circle(7)
rectangle = Rectangle(8, 5)
triangle = Triangle(10, 6)

# Circle
circle.draw()
print("Circle Area:", circle.area())

print()

# Rectangle
rectangle.draw()
print("Rectangle Area:", rectangle.area())

print()

# Triangle
triangle.draw()
print("Triangle Area:", triangle.area())
```
