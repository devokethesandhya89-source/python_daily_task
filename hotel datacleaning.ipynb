```python
#2 leading/trailing
df.columns = df.columns.str.strip()
print(df.columns)
```

```python
#3 standaralize all data formats to yyyy-mm-dd
df["Booking Date"] = pd.to_datetime( 
    df["Booking Date"], 
    errors="coerce"
)
df["Booking Date"] = df["Booking Date"].dt.strftime("%Y-%M-%d")
```

```python
# handle completely missing customer name
df["Customer Name"] =df["Customer Name"].replace(["N/A","NA",""," "],pd.NA)
print(df["Customer Name"].isnull().sum())
```

```python
#
gender_mapping ={
    "Male":"Male",
    "male":"Male",
    "MALE":"Male",
    "M":"Male",
    "Female":"Female",
    "female":"Female",
    "FEMALE":"Female",
    "F":"Female",
    "Other":"Other"
}
df["Gender"]=df["Gender"].replace(gender_mapping)
print(df["Gender"].value_counts())
print(df["Gender"].unique())
```

```python
# 5
df['City']=df['City'].str.strip().str.title()
df['City'].replace({'Hydrabad':'Hyderabad','Bengaluru':'Bangalore'})
```

```python
#6
room_mapping={
    'delux':'Deluxe','suite':'Suite',
    'standard':'Standard','exec':'Executive','STANDARD':'Standard',
}
df['Room Type']=df['Room Type'].str.strip().str.lower().map(room_mapping).fillna('Unknown')
print(df['Room Type'].value_counts())
print(df['Room Type'].unique())
```

```python
#7
channel_mapping={
    'ONLINE':'Online','online':'online','walkin':'Walk-in',
    'TraveAgent':'Travel Agent','travel agent':'Travel Agent',
    'corp':'Corporate'
}
df['Booking Channel'] =df['Booking Channel'].str.strip().str.lower().map(channel_mapping).fillna('Others')
print(df['Booking Channel'].value_counts())
print(df['Booking Channel'].unique())
```

```python
print(df['Customer Segment'].unique())
```

```python
#8
Segment_mapping={
    'premium':'Premium','corporate':'Corporate',
    'regular':'Regular','REG':'Regular', 'budget':'Budget'
}
df['Customer Segment']=df['Customer Segment'].str.strip().str.lower().map(Segment_mapping).fillna('Others')
print(df['Customer Segment'].value_counts())
print(df['Customer Segment'].unique())
```

```python
#9
df['Age']=df['Age'].astype(str).str.extract(r'(\d+)').astype(float)
df['Age'] =pd.to_numeric(df['Age'],errors = 'coerce')
df.loc[(df['Age'] < 18) |(df['Age'] > 100),'Age'] = np.nan
print(df['Age'])
```
